Round the RMSE scores to 3 decimal digits using round(score, 3). This keeps the imputation difference visible in this release.
Which option gives better RMSE?

```python
#create two versions of df
df_train_v1 = df_train.copy()
df_val_v1 = df_val.copy()
df_test_v1 = df_test.copy()

df_train_v1 = df_train_v1.reset_index(drop=True)
df_val_v1 = df_val_v1.reset_index(drop=True)
df_test_v1 = df_test_v1.reset_index(drop=True)
```

```python
df_train_v2 = df_train.copy()
df_val_v2 = df_val.copy()
df_test_v2 = df_test.copy()

df_train_v2 = df_train_v2.reset_index(drop=True)
df_val_v2 = df_val_v2.reset_index(drop=True)
df_test_v2 = df_test_v2.reset_index(drop=True)
```

```python
#set up the linear regression functions
def train_linear_regression(X, y):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)
    
    return w_full[0], w_full[1:]



#set up RMSE Function
def rmse(y, y_pred):
    se = (y - y_pred) ** 2
    mse = se.mean()
    return np.sqrt(mse)
```

```python
#set up the y value

y_train = np.log1p(df_train.fuel_efficiency_mpg.values)
y_val = np.log1p(df_val.fuel_efficiency_mpg.values)
y_test = np.log1p(df_test.fuel_efficiency_mpg.values)

```

```python
del df_train_v1['fuel_efficiency_mpg']
del df_val_v1['fuel_efficiency_mpg']
del df_test_v1['fuel_efficiency_mpg']

del df_train_v2['fuel_efficiency_mpg']
del df_val_v2['fuel_efficiency_mpg']
del df_test_v2['fuel_efficiency_mpg']
```

```python
def prepare_X(df):
    base = ['engine_displacement',
'horsepower',
'vehicle_weight']
    features = base.copy() 
    print(features)
    for name, values in categorical.items():
        for value in values:
            df['%s_%s' % (name, value)] = (df[name] == value).astype(int)
            features.append('%s_%s' % (name, value))
    print(features)
    df_num = df[features]
    df_num = df_num.fillna(0)
    X = df_num.values

    return X
```

```python
#train the model for version 1
X_train_v1 = prepare_X(df_train_v1)

w0_v1, w_v1 = train_linear_regression(X_train_v1, y_train_v1)

y_pred_v1 = w0_v1 + X_train_v1.dot(w_v1)
```

```python
#train the model for version 2
X_train_v2 = prepare_X(df_train_v2)

w0_v2, w_v2 = train_linear_regression(X_train_v2, y_train_v2)

y_pred_v2 = w0_v2 + X_train_v2.dot(w_v2)
```

Question 4
Now let's train a regularized linear regression.
For this question, fill the NAs with 0.
Try different values of r from this list: [0, 0.01, 0.1, 1, 5, 10, 100].
Use RMSE to evaluate the model on the validation dataset.
Round the RMSE scores to 4 decimal digits. This keeps the small but real regularization differences visible instead of turning several choices into a tie.
Which r gives the best RMSE?

```python
#create function to train linreg with penalty
def train_linear_regression_reg(X, y, r=0.001):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX = XTX + r * np.eye(XTX.shape[0])

    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)
    
    return w_full[0], w_full[1:]
```

```python
for r in [0, 0.01, 0.1, 1, 5, 10, 100]:
    w0, w = train_linear_regression_reg(X_train_v2, y_train, r=r)

    X_val = prepare_X(df_val_v2)
    y_pred = w0 + X_val.dot(w)
    score = rmse(y_val_v2, y_pred)
    
    print(r, w0, round(score,3))
```
